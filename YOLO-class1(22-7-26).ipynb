```python
from ultralytics import YOLO
```

```python
model = YOLO("yolo11n.pt")
```

```python
model.predict("/content/dog.png")[0].show()
```

```python
model.predict("/content/dog.png")[0].plot()
```

```python
from ultralytics import YOLO
import gradio as gr
!pip install gradio
```

```python
def detect_objects(image):
    results = model(image)

    # Annotated Image
    annotated_image = results[0].plot()

    return annotated_image
```

```python
app = gr.Interface(    fn=detect_objects,
    inputs=gr.Image(type="numpy"),
    outputs=gr.Image(type="numpy"),)
```

```python
app.launch(show_error=True)
```
